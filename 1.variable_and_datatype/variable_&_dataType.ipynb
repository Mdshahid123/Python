##  2)Rules for writing variables

### 1) variable name  can be only alphabet
### 2) variable name can be only underscore
### 3) varaibele name can be any combination of underscore,numeric and alphabet character but cannot start with numeric character 
### 4) and cannot be any python keyword



```python
# example 1:
# varaible name can be only alphabet
a=5
abd=9
print(a,abd)
```

```python
# example:02
# variable name can be only underscore
_=3
__=6
print(_,__)



```

```python
# example-03
# variable name can be any combination of numeric,underscore and alphabet but cannot be start with numeric character
my_variable=5
hello_abc=6
print(my_variable,hello_abc)

```

```python
# example-04 
# variable name cannot be any python reserved keyword
a=3
print(a)
```

```python
# variable names are  case-sensitive (age,AGE and AgE id diffrent vaiable)
age=6
Age=8
AGE=9
print(age,AGE,Age)

```
