### #Dictionary is a predefined data type  used to store collection of  labeled data in  key:value pairs.

## properties of dictinory

#### It is a hetro,mutable,ordered and don not allow duplicates 

```python
# creating a dictinary with data
student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}

print(student)


# clearly it is order  bcz it maintain the order in which it insert 
# it can store hetorgenius data type 

```

```python
# creating a emty dict and later on we filled it 

student={}

student["name"]="md shahid"
student["age"]="24"
student["isstudent"]=False


print(student)


```

```python
# accessing the dict element or items  without loop 
student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}
print(student)
print(student["name"])
print(student["age"])
print(student["isStudent"])


```

```python
# eccessing the dict elemnts using loop
# we can use range here insetad we will be use list collection

student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}



for key in student :
  print(key)  # here  we will get a key 
  print(student[key])   # here we will eccessing a valu using that key
  

```

```python
student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}

for key in student.keys():
  print(key)
```

```python
student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}

for value in student.values():
  print(value)
```

```python
student={
   "name":"md shahid",
   "age":24,
   "isStudent":True
}

for item in student.items():
  print(item)
```

```python
# dict doesn't allow duplicate key that  means keys aloways should be unique

student={
  "name":"md shaid",
  "age":24,
  "name":"md"

}
print(student)


# value of name will be updated with the last name value 
```

```python
student={
  "age":24,
  "name":"md",
  "name":"student",
  "name":"shristi"

}
print(student)
```

### #dictinary is mutable that means we can modifi the dictinary after creating it  that is we can change the items values ,we can add new items or we we can remove the items 

```python
student={
  "age":24,
  "name":"md",
 

}

# let's change the value of dict
student["age"]=45
student["name"]="shahid"

print(student)
```

```python
# let's add new items 

student={
  "age":24,
  "name":"md",
 

}

student["isStudent"]=True

# here dictinary is growing  bcz we hare adding new items 
```

```python
# let's remove the items from the list

student={
  "age":24,
  "name":"md",
 

}

del student["age"]
print(student)

# here dict is shrinking bcz we are removing the imtes from the dict
```

```python
student={
  "age":24,
  "name":"md",
 

}

student.clear()
print(student)
```
