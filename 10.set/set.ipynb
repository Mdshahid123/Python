## What is a Set?
#### A set is a predefined Python data type used to store a collection of unique unlabbeld data

## Key characteristics of sets
#### 1. Unordered: Items do not have a fixed position.
#### 2. No duplicates: Repeated values are stored only once.
#### 3. Mutable: You can add or remove items after the set is created.
#### 4. Heterogeneous: A set can contain different immutable data types.
#### 5. No indexing: Because sets are unordered, my_set[0] is not allowed.
#### 6. Iterable: We can use a loop to visit each item.

## 1)Creating a set

```python
# Creating sets with elements
fruits = {"apple", "banana", "apple", "orange"}
numbers = {56, 78, 99, 100, 98}
mixed_set = {24, "shahid", 4.5, True}

# The duplicate apple is automatically removed
print(fruits)
print(numbers)
print(mixed_set)


# clealry above example showing that 
# we can store homo and hetro data in set 
# uinique data 
#  it is  unorder 
```

## 2)Accessing set items
#### Sets do not support indexes or slices. Use a loop or membership operator instead.

```python
colors = {"red", "green", "blue"}

print(colors[0])

# Accessing items using a loop
for color in colors:
    print(color)

# Membership operator
if "green" in colors:
    print("green is present")
```

## 3)Sets are mutable
#### We cannot change one item by index, but we can add and remove whole items.

```python
# Adding one item
fruits = {"banana", "apple"}
fruits[0]="mango"
print(fruits)

fruits.add("orange")  # it will add the item
print(fruits)

# Adding multiple items
fruits.update(["mango", "grapes", "apple"])
print(fruits)  # apple is not added again
```

```python
# Removing items
fruits = {"banana", "apple", "orange"}
fruits.remove("apple")  # gives an error if the item is absent
print(fruits)

fruits.discard("mango")  # no error if mango is absent
print(fruits)

removed_item = fruits.pop()  # removes an arbitrary item
print("Removed:", removed_item) 
print(fruits)
```

```python
# Clearing all items
numbers = {10, 30, 1, 8}
numbers.clear()
print(numbers)
```

## 4) Set operations
#### Set operations are useful when comparing collections of unique values.

```python
a = {1, 2, 3, 4}
b = {3, 4, 5, 6}

print("Union:", a | b)              # all unique items
print("Intersection:", a & b)       # common items
print("Difference a - b:", a - b)  # items only in a
print("Difference a - b:", b - a)  # items only in b
print("Symmetric difference:", a ^ b) # it will give items which is  not common to both
```

```python
# Converting a list to a set removes duplicates
marks = [80, 75, 80, 90, 75, 95]
unique_marks = set(marks)
print(unique_marks)

# sorted() returns a new sorted list
print(sorted(unique_marks)) # it will sort the given set and return the list
```
