#### Type casting in python
#### Type casting means converting one data type into another data type.
#### type cating can be implict or explicit

##### let's see with the real situation where we need a type casting
##### situation-01

```python
# take two number as a  input from the  user and add it 

num1,num2=input("enter two number").split(" ")
print(int(num1)+int(num2))


# here we are doing explicit converstion
```
