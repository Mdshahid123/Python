## now printing the msg with f string 
## string with external variable or expresstion is called f string 
## The process of inserting an external variable or expression into a string is called string interpolation.

```python
a=10
b=5
c=2

sum=a+b+c

print(f"sum of {a}, {b},and {c} is {sum} ")
print(f"sum of {a}, {b},and {c} is {a+b+c} ")
```
