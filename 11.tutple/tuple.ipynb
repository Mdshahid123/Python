##  What is a Tuple?
### A tuple is a predefined Python data type used to store an ordered collection of unlabbeled items.
### Key characteristics of tuples
1. Ordered: Items keep the order in which they were inserted.
2. Immutable: We cannot change, add, or remove items after creation.
3. Allows duplicates: The same value can appear many times.
4. Heterogeneous: A tuple can hold different data types.
5. Indexing and slicing: Items can be accessed by position.
6. Iterable: We can use a loop to visit each item.
Tuples are useful for data that should not be changed, such as coordinates or fixed records.

## Creating a tuple

```python
# Creating tuples with elements
tuple1 = ("apple", "banana", "apple", "orange")
tuple2 = (56, 78, 99, 100, 98)
tuple3 = (45.6, 67.9, 87.1)
tuple4 = (24, "shahid", 4.5, True)

print(tuple1, tuple2, tuple3, tuple4)
```

```python
# Empty tuple
empty_tuple = ()
print(empty_tuple)
```

# creating a sime item in a set

```python
# A single-item tuple needs a comma
one_item = ("apple",)
not_a_tuple = ("apple")  # it will string it we don't use coma 

print(type(one_item))
print(type(not_a_tuple))
```

## accessing a elemnt of tuple

```python
items = (2, 4.5, "shahid", True)

print(items)       # whole tuple
print(items[0])    # first item
print(items[2])    # third item
print(items[-1])   # last item
print(items[1:3])  # slicing

# Accessing a nested tuple
data = (23, 45, (46, 89), (76, 89, 56))
print(data[2][1])
print(data[3][0])
```

## Tuple methods
## Because tuples are immutable, they have only two built-in methods.

```python
numbers = (45, 6, 7, 8, 7, 9, 7)

print(numbers.count(7))  # number of times 7 is present
print(numbers.index(7))  # index of first occurrence of 7
```

```python
# Converting between list and tuple


tuple=(7,8,9,10)

# Convert a tuple to a list if changes are needed
list = list(tuple)
print(list)
```

## packing and unpacking 

```python
# Packing and unpacking
student = ("Shahid", 24, "Python")  # packing
name, age, course = student              # unpacking

print(name)
print(age)
print(course)
```
