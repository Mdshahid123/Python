## What is List
#### -List is a predefined data type in python used to store a ordered,mutable  collection of homogenius and  hetrogenius data type.
#### -list can has duplicate value.
#### -list is a iterable object
#### -list is implemented as dynamic array of pointer


```python
# creating a List with elements 

list1=["apple","banana","apple","orange"] # list1 is a variable pointing to the object that store  list data type (homgenius)

list2=[56,78,99,100,98] # list2 is  a variable pointing to the object that storing list data type (homogienius )

list3=[45.6,67.9,87.10] # list3 a variable pointing to the object that store  list of  float(homgenius)

list4=[24,"shahid",4.5,True] # list4 is a variable pointing to the object that store list data type (hetro)

print(list1,list2,list3,list4)
```

## List is Mutable 
## that means we can change, add, and remove items in a list after it has been created.

```python
# let's change or modifying the items  in the  existing list
list=[2,4.5,"shahid",True]

print(f"before chnaging:{list}")
list[0]=7
list[1]=9
list[2]="sanjay"
list[3]=False
print(f"after chnaging:{list}")
```

```python
# adding a new items  at any index 
list=[2,4.5,"shahid",True]

print(f"before inserting list length is:",len(list))


# insert():it is used to add a elment at any index in existing list

list.insert(4,"sanjay")

print(f"after inserting list length is:",len(list))

# insert function can insert a new element at any index to the existing list cause a list grow
```

```python
#adding a new item at the end  to the existing list cause a list grow
fruits=["banana","apple"]
fruits.append("orange")
print(fruits)
```
