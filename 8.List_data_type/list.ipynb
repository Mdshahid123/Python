## What is List
#### -List is a predefined data type in python used to store a collection of unlabeled data 


### #Key Characteristics of Lists
  #### 1)Ordered: Items maintain a defined order based on how they were inserted
  #### 2)Mutable: You can change, add, and remove items after the list is created.
  #### 3)Allow Duplicates
  #### 4)Heterogeneous: A single list can hold mixed data types, like integers, strings, and booleans all at onc
  #### 5)Indexing:items are acceessing thorugh the index


  #### 5)list is a iterable object
  #### 6)list is implemented as dynamic array of pointer

 ## accessing a list items 

```python
# accessing a List

list=[2,4.5,"shahid",True]

# accessing a whole list 
print(list)


# accessing a indivisual elements  or items 
print(list[0])
print(list[1])
print(list[2])
print(list[3])


# accsessing a nested list(list inside a list) item
l1=[1,2,3,4,[89,87]]
print(l1)
print(l1[4])
print(l1[4][0])



list=[23,45,67,[46,89],[76,89,56]]
print(list[3][1])
print(list[3][0])
print(list[4][1])



l1=[23,45,67,89,[23,45,67],[[23,45],[35,67]]]
print(l1[4][1])
print(l1[5][1][0])

```

```python
# accessing a list using loop

list=[23,34,56,7,88,9]
for index in range(0,6,1):
  print(list[index])

   # or (better way)
   
length=len(list)
for index in range(0,length,1):
  print(list[index])

```

```python
# we have alternate way to apply the loop on list without using range()

l1=[23,45,67,8,9]
for item in l1:
  print(item)


  # or 

for item in [23,45,67,8,9,5,6,7,7,89,9,0,8]:
  print(item)
```

## List is Mutable 
## that means we can change, add, and remove items in a list after it has been created.

```python
# let's change or modifying the items  in the  existing list
list=[2,4.5,"shahid",True]

print(f"before chnaging:{list}")
list[0]=7
list[1]=9
list[2]="sanjay"
list[3]=False
print(f"after chnaging:{list}")
```

```python
# adding a new items  at any index 
list=[2,4.5,"shahid",True]

print(f"before inserting the list length is:",len(list))
print(list)


# insert():it is used to add a elment at any index in existing list

# list.insert(0,"sanjay")
# list.insert(2,"sanjay")
# list.insert(4,"sanjay")

print(f"after inserting list length is:",len(list))
print(list)

# insert function can insert a new element at any index to the existing list cause a list grow
```

```python
#adding a new item at the end  to the existing list cause a list grow
fruits=["banana","apple"]
fruits.append("orange")
print(fruits)
```

```python
#extend function can be used to merge two list cauing a list grow
fruits1=["banana","apple"]
fruits2=["mango","grapes"]

fruits1.extend(fruits2) # merge [fruits1,fruits2]

print(fruits1)
print(fruits2)

```
