## taking input from the user

```python
#take name  as a input from the the user and print this
name=input("enter you name")
print(name)
```

```python
# taking a multiple string  name,college,gmail and print the msg with value
name=input("enter you name ")
age =input("enter you  age ")
college =input("enter you  college ")

print(f"your name ,college and age are : {name},{college},{age}")
```

```python
#taking multiple string in one input
name,age,college=input("enter your name ,age and college").split(" ")
print(f"your name ,college and age are : {name},{college},{age}")
```

```python
# taking a five number  on one input
a=input("enter 5 number ") # here user input values wil be store in a variable as a string "num1 num2 num3 num4 num5 "
print(a)



# problem here is that we have to take a five number but actually we assigning that five number as  string 
```

```python
# instead creating multiple variable we can use single variable to store those value

nums=list(map(int,input("enter a five number").split()))
print(nums)

```
