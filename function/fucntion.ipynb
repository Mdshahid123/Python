## diffrent way to perform a task through the fucntion 

### 1) without parameters and return nothing 

```python
# defined the function to calcuate the addtion 

# defined 
def addtion_of_two_number():
  a=2
  b=3
  sum=a+b
  print(sum)

# lt'es call the fucntion
addtion_of_two_number()
```

### 2) with parameters and return nothing 

```python
def addtion_of_two_number(a,b):
  sum=a+b
  print(sum)

# lt'es call the fucntion
addtion_of_two_number(2,3)


# argument:when we call the fucntion by passing the  values then that values is called argument 
# paramter: parameter is a variable inside the parenthesis to receive a argument 
```

### 3) with parameters and return something 

```python
def addtion_of_two_number(a,b):
  sum=a+b
  return sum

# lt'es call the fucntion
sum=addtion_of_two_number(2,3)
print(sum)
```

### 4) without any paramter  and return something 

```python
def addtion_of_two_number():
  a=2
  b=3
  sum=a+b
  return sum

# lt'es call the fucntion
sum=addtion_of_two_number()
print(sum)
```

## default argument 

```python
def addtion_of_two_number(a,b=7):
  sum=a+b
  return sum

# lt'es call the fucntion
sum1=addtion_of_two_number(2,6) #here passed argument will be used for paramter b
sum2=addtion_of_two_number(2)  # here default value will be used for paramter b
print(sum1)
print(sum2)


# deafult argument:when user dont pass the corressponding argument of recving parmater then it's default value will be use otherwise passed argument will be use 





```
